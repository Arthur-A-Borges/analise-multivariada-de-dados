Importando biblioteca "pandas" e organizando o DataSet

```python
import pandas as pd

#Utilizei o sep, para o pandas verificar que a separação é ";"
df = pd.read_csv("cidades_vacinacao.csv", encoding = "utf-8", sep = ";")

df.head()
```

Limpando as colunas vazias

```python
df = df.dropna(axis=1, how='all')
```

```python
df.head()
```

Verificando informações do DataSet

```python
df.info()
```

Alterando "Data de notificação" para o Tipo "Date"

```python
df['Data da Notificação'] = pd.to_datetime(df['Data da Notificação'], dayfirst=False, errors ="coerce")

print(df.dtypes)
```

Alterando "Idade de Evento" para o tipo "Inteiro"

```python
df['Idade Evento'] = pd.to_numeric(df['Idade Evento'], errors='coerce')

print(df.dtypes)
```

```python
df.head()
```

Salvando planilha limpa

```python
df.to_csv("cidade_vacinacao_limpa.csv")
```
